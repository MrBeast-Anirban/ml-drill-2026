```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/kaggle/input/datasets/sumanthvrao/daily-climate-time-series-data/DailyDelhiClimateTest.csv')
```

```python
df = df[['date', 'meantemp']]
```

```python
df.head()
```

```python
plt.scatter(df['date'], df['meantemp'], color = 'red')
```

```python
x1 = df['meantemp'].ewm(alpha = 0.9).mean()
```

```python
x1
```

```python
df['ewma'] = x1
df.head()
```

```python
plt.scatter(df['date'], df['meantemp'], color = 'red')
plt.plot(df['date'], x1, color = 'black', label = 'ewma alpha = 0.9')
plt.legend()
plt.show()
```

```python
# ewma with alpha = 0.7
x2 = df['meantemp'].ewm(alpha = 0.7).mean()

plt.scatter(df['date'], df['meantemp'], color = 'red')
plt.plot(df['date'], x2, color = 'black', label = 'ewma alpha = 0.7')
plt.legend()
plt.show()
```

```python
# ewma with alpha = 0.5
x3 = df['meantemp'].ewm(alpha = 0.5).mean()

plt.scatter(df['date'], df['meantemp'], color = 'red')
plt.plot(df['date'], x3, color = 'black', label = 'ewma alpha = 0.5')
plt.legend()
plt.show()
```
