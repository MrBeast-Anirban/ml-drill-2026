## Branching

```python
# Inputs with 3 columns
x = Input(shape = (3, ))
```

```python
hidden1 = Dense(128, activation = 'relu')(x)
hidden2 = Dense(128, activation = 'relu')(hidden1)

#Branching
# Branch 1 --> Age Prediction, Branch 2 --> Place Prediction
output1 = Dense(1, activation = 'linear')(hidden2) # age prediction
output2 = Dense(1, activation = 'sigmoid')(hidden2) #delhi/mumbai
```
