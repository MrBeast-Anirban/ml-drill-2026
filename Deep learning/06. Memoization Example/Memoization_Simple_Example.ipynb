```python
# Fibonacci without memoization
```

```python
def fib(n):
  if n == 1 or n == 0:
    return 1
  else:
    return fib(n-1) + fib(n-2)
```

```python
import time
start = time.time()
fib(40)
print('Time taken :- ', time.time()-start)
```

```python
# Fibonacci with memoization
```

```python
def fib(n, d):
  if n in d:
    return d[n]
  else:
    d[n] = fib(n-1, d) + fib(n-2, d)
    return d[n]
```

```python
start = time.time()
d = {0:1, 1:1}
fib(40, d)
print('Time taken :- ', time.time()-start)
```
