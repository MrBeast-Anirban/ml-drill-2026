```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="darkgrid", context="talk", palette="muted")
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
titanic = pd.read_csv("train.csv")
```

```python
titanic.head()
```

```python
flights = sns.load_dataset("flights")
```

```python
flights.head()
```

```python
iris = sns.load_dataset("iris")
```

```python
iris.head()
```

## 1. Scatterplot (Numerical - Numerical)

```python
sns.scatterplot(x = tips['total_bill'], y = tips['tip'])
```

```python
# multivariate
sns.scatterplot(x = tips['total_bill'],y = tips['tip'],hue=tips['sex'],style=tips['smoker'],size=tips['size'])
```

## 2. Bar Plot (Numerical - Categorical)

```python
sns.barplot(x = titanic['Pclass'],y = titanic['Age'])
```

```python
# multivariate
sns.barplot(x = titanic['Pclass'],y = titanic['Age'], hue = titanic["Sex"])
```

## 3. Box Plot (Numerical - Categorical)

```python
sns.boxplot(x = titanic['Sex'], y = titanic['Age'])
```

```python
# multivariate
sns.boxplot(x = titanic['Sex'], y = titanic['Age'], hue = titanic['Survived'])
```

```python
sns.distplot(titanic[titanic['Survived']==0]['Age'], hist = False)
sns.distplot(titanic[titanic['Survived']==1]['Age'], hist = False)
```

## 4. Heatmap (Catgorical - Categorical)

```python
crosstab = pd.crosstab(titanic['Pclass'], titanic['Survived'])
crosstab
```

```python
sns.heatmap(crosstab)
```

## 6. Clustermap (Catgorical - Categorical)

```python
sns.clustermap(pd.crosstab(titanic['Parch'], titanic['Survived']))
```

## 7. Pairplot

```python
sns.pairplot(iris, hue = 'species')
```

## 8. Lineplot (Numerical - Numerical)

Used for timestamp data to represent relation with respect to time.

```python
flights.head()
```

```python
flights_new = flights[['year', 'passengers']]
flights_new.groupby('year').sum().reset_index()
```

```python
sns.lineplot(x = flights_new['year'], y = flights_new['passengers'])
```

```python
sns.clustermap(flights.pivot_table(values='passengers',index='month',columns='year'))
```
