## 1. Working with Dates

```python
# converting to datetime datatype from object
date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

### 1. Extract year


```python
date['date_year'] = date['date'].dt.year
date.sample(2)
```

### 2. Extract Month


```python
date['date_month_no'] = date['date'].dt.month
date.sample(2)
```

```python
date['date_month_name'] = date['date'].dt.month_name()
date.sample(2)
```

### 3. Extract Days


```python
date['date_day'] = date['date'].dt.day
date.sample()
```

### 4. Date of week

```python
date['date_dow'] = date['date'].dt.dayofweek
date.sample(2)
```

```python
date['date_dow_name'] = date['date'].dt.day_name()
date.sample(2)
```

```python
date.drop(columns = ['product_id', 'city_id', 'orders']).head()
```

### 5. Check if weekend

```python
date['date_is_weekend'] = np.where(date['date_dow_name'].isin(['Sunday', 'Saturday']), 1, 0)

date.drop(columns=['product_id','city_id','orders']).head()
```

### 6. Extract week of the year

```python
date['date_week'] = date['date'].dt.isocalendar().week
date.drop(columns=['product_id','city_id','orders']).sample(2)
```

### 7. Extract Quarter

```python
date['quarter'] = date['date'].dt.quarter

date.drop(columns=['product_id','city_id','orders']).sample(2)
```

### 8. Extract Semester

```python
date['semester'] = np.where(date['quarter'].isin([1,2]), 1, 2)

date.drop(columns=['product_id','city_id','orders']).sample(2)
```

### 9. Extract Time elapsed between Dates

```python
import datetime
today = datetime.datetime.today()
today
```

```python
today - date['date']
```

```python
# days passed
(today-date['date']).dt.days
```

```python
# Months passed (Not supporting)
np.round((today -date['date']) / np.timedelta64(1, 'M'),0)
```

## 2. Working with Time

```python
time.sample(2)
```

```python
time.info()
```

```python
# Converting to datetime datatype
time['date'] = pd.to_datetime(time['date'])
```

```python
time['hours'] = time['date'].dt.hour
time['minutes'] = time['date'].dt.minute
time['seconds'] = time['date'].dt.second

time.head()
```

### Extract Time Part

```python
time['time'] = time['date'].dt.time

time.head()
```

### Time Difference

```python
today - time['date']
```

```python
# in seconds

(today - time['date'])/np.timedelta64(1, 's')
```

```python
# in minutes
(today - time['date'])/np.timedelta64(1, 'm')
```

```python
# in hours
(today - time['date'])/np.timedelta64(1, 'h')
```
