# Analytical closed-form solution for simple linear regression

\begin{aligned}
m &= \frac{\sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})}
{\sum_{i=1}^{n} (x_i - \bar{x})^2} \\
b &= \bar{y} - m\bar{x}
\end{aligned}



# Complete OLS Derivation for Simple Linear Regression

### 1. Define the Objective Function
We start by minimizing the Residual Sum of Squares ($RSS$), which aggregates the squared differences between observed values $y_i$ and predicted values $\hat{y}_i = mx_i + b$.

$$
\begin{aligned}
S(m, b) &= \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 \\
&= \sum_{i=1}^{n} (y_i - (mx_i + b))^2
\end{aligned}
$$

---

### 2. Derive the Intercept ($b$)
To minimize the error function with respect to $b$, we take the partial derivative, set it equal to zero, and isolate $b$.

$$
\begin{aligned}
\frac{\partial S}{\partial b} &= \frac{\partial}{\partial b} \sum_{i=1}^{n} (y_i - mx_i - b)^2 \\
&= -2 \sum_{i=1}^{n} (y_i - mx_i - b) = 0
\end{aligned}
$$

Divide the entire equation by $-2$:
$$ \sum_{i=1}^{n} (y_i - mx_i - b) = 0 $$

Distribute the summation across the terms:
$$ \sum_{i=1}^{n} y_i - m \sum_{i=1}^{n} x_i - \sum_{i=1}^{n} b = 0 $$

Since summing a constant $b$ from $1$ to $n$ yields $\sum_{i=1}^{n} b = nb$:
$$ \sum_{i=1}^{n} y_i - m \sum_{i=1}^{n} x_i - nb = 0 $$

Isolate $nb$ and divide the entire equation by $n$:
$$
\begin{aligned}
nb &= \sum_{i=1}^{n} y_i - m \sum_{i=1}^{n} x_i \\
b &= \frac{1}{n} \sum_{i=1}^{n} y_i - m \left( \frac{1}{n} \sum_{i=1}^{n} x_i \right)
\end{aligned}
$$

Substitute the definitions of the sample means $\bar{y} = \frac{1}{n} \sum_{i=1}^{n} y_i$ and $\bar{x} = \frac{1}{n} \sum_{i=1}^{n} x_i$:
$$ b = \bar{y} - m\bar{x} $$

---

### 3. Derive the Slope ($m$)
Next, we minimize the error function with respect to $m$ by taking the partial derivative and setting it to zero.

$$
\begin{aligned}
\frac{\partial S}{\partial m} &= \frac{\partial}{\partial m} \sum_{i=1}^{n} (y_i - mx_i - b)^2 \\
&= -2 \sum_{i=1}^{n} x_i (y_i - mx_i - b) = 0
\end{aligned}
$$

Divide by $-2$ and substitute our derived expression for $b$ ($b = \bar{y} - m\bar{x}$):
$$
\begin{aligned}
\sum_{i=1}^{n} x_i (y_i - mx_i - (\bar{y} - m\bar{x})) &= 0 \\
\sum_{i=1}^{n} x_i ((y_i - \bar{y}) - m(x_i - \bar{x})) &= 0
\end{aligned}
$$

Distribute $x_i$ inside the parenthetical terms:
$$ \sum_{i=1}^{n} x_i (y_i - \bar{y}) - m \sum_{i=1}^{n} x_i (x_i - \bar{x}) = 0 $$

Move the second summation term to the right-hand side:
$$ \sum_{i=1}^{n} x_i (y_i - \bar{y}) = m \sum_{i=1}^{n} x_i (x_i - \bar{x}) $$

Apply standard mathematical mean-deviation identity adjustments, where $\sum_{i=1}^{n} x_i(y_i - \bar{y}) = \sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})$ and $\sum_{i=1}^{n} x_i(x_i - \bar{x}) = \sum_{i=1}^{n} (x_i - \bar{x})^2$:
$$ \sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y}) = m \sum_{i=1}^{n} (x_i - \bar{x})^2 $$

Isolate $m$ to yield the final closed-form slope formula:
$$ m = \frac{\sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})}{\sum_{i=1}^{n} (x_i - \bar{x})^2} $$


```python
class selflr:
  def __init__(self):
    self.m = None
    self.b = None
  def fit(self, X_train, y_train):
    num = 0
    den = 0
    for i in range (X_train.shape[0]):
      num = num + ((X_train[i] - X_train.mean())*(y_train[i] - y_train.mean()))
      den = den + ((X_train[i] - X_train.mean())*(X_train[i] - X_train.mean()))
    self.m = num/den
    self.b = y_train.mean() - (self.m * X_train.mean())
    print(self.m)
    print(self.b)
  def predict(self, X_test):
    return self.m * X_test + self.b
```
