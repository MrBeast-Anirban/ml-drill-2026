```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="darkgrid", context="talk", palette="muted")
```

```python
df = pd.read_csv("train.csv")
```

```python
df.head()
```

## 1. Categorical Data

### a. Count plot


```python
#sns.countplot(df["Embarked"])
sns.countplot(df, x = df["Embarked"]) # just to rotate the plot
```

### b. Pie Chart

```python
df['Sex'].value_counts().plot(kind = "pie", autopct = "%.2f")
```

## 2. Numerical Data

### a. Histogram

```python
plt.hist(df["Age"], bins=5)
```

### b. Distplot

```python
sns.distplot(df["Age"])
```

### c. Box Plot

```python
sns.boxplot(df["Age"])
```
