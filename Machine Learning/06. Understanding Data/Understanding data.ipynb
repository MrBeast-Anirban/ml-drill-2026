# 7 questions always to ask to understand the data.
1. How big is the data?
2. How does the data look like?
3. What is the datatype of the column?
4. Are there any missing values?
5. How does the data look mathematically?
6. Are there duplicate values?
7. How is the correlation between columns?

```python
import pandas as pd
```

```python
df = pd.read_csv('train.csv')
```

## 1. How big is the data?

```python
df.shape
```

## 2. How does the data look like?

```python
# alternative to df.head() but he rows are provided randomly
df.sample(5)
```

## 3. What is the data type of cols?

```python
# helps to check the data type and reduce the size if required.
df.info()
```

## 4. Are there any missing values?

```python
# can see "Age" and "Cabin" column has some null values
df.isnull().sum()
```

## 5. How does the data look mathematically?

```python
df.describe()
```

## 6. Are there duplicated values?

```python
df.duplicated().sum()
```

## 7. How is the correlation between columns?

```python
df.corr(numeric_only=True)
```

```python
df.corr(numeric_only=True)["Survived"]
```
