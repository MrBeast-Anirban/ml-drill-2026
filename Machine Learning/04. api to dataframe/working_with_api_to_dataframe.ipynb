```python
import pandas as pd
import requests # create http request
```

```python
# request the server to fetch details using api key
response = requests.get('https://api.themoviedb.org/3/movie/top_rated?api_key=7fbc7861cb903f427b0dfbe8f72a6be3&language=en-US&page=1')
```

```python
response.json()
```

```python
# we only want the results part
response.json()['results']
```

```python
# convert the list response.json()['results'] to dataframe
pd.DataFrame(response.json()['results']).head(2)
```

```python
# extract selective columns and assign to temp_df
temp_df = pd.DataFrame(response.json()['results'])[['id', 'title', 'overview', 'release_date', 'popularity', 'vote_average', 'vote_count']]
```

```python
temp_df.head()
```

```python
# in response.json() output you can see that there are 549 pages, and we are at page 1 currently
# so we will iterate through all these pages and append them to the older dataframe

df = pd.DataFrame() # creating empty dataframe
df
```

```python
#a = response.json()['total_pages'] # get total pages at page 501 getting error
for i in range (1,500):
  response = requests.get('https://api.themoviedb.org/3/movie/top_rated?api_key=7fbc7861cb903f427b0dfbe8f72a6be3&language=en-US&page={}'.format(i))
  temp_df = pd.DataFrame(response.json()['results'])[['id', 'title', 'overview', 'release_date', 'popularity', 'vote_average', 'vote_count']]
  print("Results from Page {} extracted.".format(i))
  df = pd.concat([df, temp_df], ignore_index=True) # to avoide the same index in every iteration concatenated (Note: df.append() method doesnt work) append method was completely removed in Pandas 2.0
```
